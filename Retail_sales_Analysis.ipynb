# Part 1: Initializing and Loading Dataset


```python
import pandas as pd

df = pd.read_excel('Store.xls')
print(f"Dataset Shape: {df.shape}")
```

# Part 2: Extracting Column Schema Data Types


```python
print(df.dtypes)
```

# Part 3: Generating Universal Descriptive Statistics


```python
df.describe(include='all')
```

# Part 4: Row and Column Slicing via Positional Matrix Indexing


```python
exact_cell = df.iloc[0, 4]
print(f"Value at: {exact_cell}\n")

df.iloc[:5, :3]
```

# Part 5: Filtering Target Characters Using String Prefixes


```python
customers_starting_A = df[df["Customer Name"].str.startswith("A", na=False)]
print(len(customers_starting_A))
customers_starting_A.head()

```

# Part 6: Identifying Corporate Entities with String Suffixes


```python
corporate_inc = df[df["Customer Name"].str.endswith("Inc.", na=False)]
print(len(corporate_inc))
corporate_inc.head()

```

# Part 7: Multi-Prefix Subsetting on Inventory Classifications


```python
target_prefixes = ("OFF-", "TEC-")
prefix_filtered_df = df[df["Product ID"].str.startswith(target_prefixes, na=False)]
print(len(prefix_filtered_df))
prefix_filtered_df.head()

```

# Part 8: Computing Numerical Extremes for Financial Audit


```python
highest_sale = df["Sales"].max()
lowest_profit = df["Profit"].min()

print(highest_sale)
print(lowest_profit)
```
